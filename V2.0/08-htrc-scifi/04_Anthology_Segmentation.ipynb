# Anthology Segmentation

The expanded workset includes 918 anthology volumes -- collections containing multiple stories by different authors. For text analysis, anthologies present a challenge: the EF API treats each volume as a single unit, but the actual content is a mix of distinct works.

This notebook explores approaches to segmenting anthology volumes into their constituent stories using:

1. **Page-level topic shifts** -- detect boundaries where the vocabulary changes dramatically
2. **Token diversity windows** -- sliding window analysis of vocabulary turnover
3. **Structural signals** -- pages with very few tokens (title pages, blank pages between stories)

## Data

- 918 anthology volumes from `sf-hathitrust-anthologies.tsv`
- EF API page-level token counts (body/header/footer separated)
- Page sequence numbers for ordering
